# Nonlinear dynamics according to Strogatz

## 3 Bifurcations

### 3.1 Saddle node (fold) bifurcations

Prototype
$$
\dot x = r+x^2
$$

S***addle-node (fold) bifurcations is the basic mechanism by which fixed points are created and destroyed***

* $r<0$: One stable and one unstable fixed point
* $r=0$: One half-stable fixed point
* $r>0$: No fixed points

FIxed point:
$$
x = \pm \sqrt{-r}
$$

```python
r = np.linspace(-10,0, 100)
x1 = np.sqrt(-r)
x2 = -np.sqrt(-r)
```

```python
plt.close('all')

fig, ax = plt.subplots()
ax.set_title('Saddle-node bifurcation')
ax.plot(r, x1, '--k')
ax.plot(r, x2, '-k')
ax.grid('True')
plt.show()
```

### 3.2 Transcritical bifurcation

Normal form:
$$
\dot x = rx-x^2
$$

Fixed points are at $x=0$ and $x=r$. For $r<0$ the zero is stable, for $r>0$ the zero is unstable.

```python
plt.close('all')
fig, ax = plt.subplots()
ax.set_title('Transcritical bifurcation')
ax.plot([-2,0], [0,0], 'k')
ax.plot([-2,0], [-2,0], '--k')
ax.plot([0,2], [0,2], 'k')
ax.plot([0,2], [0,0], '--k')
ax.grid('True')
plt.show()
```
